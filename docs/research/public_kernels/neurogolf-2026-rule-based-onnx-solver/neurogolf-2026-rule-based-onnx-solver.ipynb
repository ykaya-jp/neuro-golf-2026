## Competition Overview
 
`max(1, 25 − ln(params + bytes + MACs))` — smaller networks score higher.
 
| Architecture | Cost | Score |
|---|---|---|
| Identity / Slice/Pad (cost=0) | 0 | **25.0** |
| Channel Gather | ~36,000–45,000 | **14.5** |
| Tiny logic (Mul, ReduceMax) | ~40,000–100,000 | **13–14.5** |
| Small CNN | ~130,000–450,000 | **11.5–13** |
| Dynamic shapes / banned ops | any | **0** |
 
**Static shapes `[1,10,30,30]` required.** Dynamic shapes, Loop, Scan, NonZero, Unique are banned (score 0).

````python
import os, math, zipfile, shutil
from pathlib import Path
from collections import Counter

import numpy as np

try:
    import onnx
    from onnx import numpy_helper
    import onnxruntime as ort
except ImportError:
    os.system("pip install -q onnx onnxruntime")
    import onnx
    from onnx import numpy_helper
    import onnxruntime as ort

ort.set_default_logger_severity(3)

# ── Paths ──────────────────────────────────────────────────────────────────
SOURCE_ZIPS = {
    "sub4979_07": Path("/kaggle/input/datasets/jonathanchan/sub4979-07"),
    "sub4971_98": Path("/kaggle/input/datasets/jonathanchan/sub4971-98"),
}
OUT_ZIP   = Path("/kaggle/working/submission.zip")
MAX_BYTES = int(1.44 * 1024 * 1024)
BANNED    = {"Loop","Scan","NonZero","Unique"}
NUM_TASKS = 400

print("=" * 55)
print("  NeuroGolf 2026 — Submission Builder")
print("=" * 55)

# ── Cost calculator ────────────────────────────────────────
def compute_cost(raw: bytes):
    if len(raw) > MAX_BYTES: return None
    try:
        m = onnx.load_model_from_string(raw)
        m = onnx.shape_inference.infer_shapes(m)
    except: return None
    tb={1:4,6:4,7:8,10:2,11:8,16:2,2:1,3:1,4:2,5:2}
    params=nbytes=macs=0; shapes={}; ops=set()
    for init in m.graph.initializer:
        dims=list(init.dims); p=math.prod(dims) if dims else 1
        params+=p; nbytes+=p*tb.get(init.data_type,4); shapes[init.name]=dims
    for inp in m.graph.input:
        s=[]
        for d in inp.type.tensor_type.shape.dim:
            if d.HasField('dim_value'): s.append(d.dim_value)
            else: return None
        shapes[inp.name]=s
    for vi in list(m.graph.value_info)+list(m.graph.output):
        s=[]
        for d in vi.type.tensor_type.shape.dim:
            if d.HasField('dim_value'): s.append(d.dim_value)
            else: return None
        if s:
            p=math.prod(s); nbytes+=p*tb.get(vi.type.tensor_type.elem_type,4); shapes[vi.name]=s
    for node in m.graph.node:
        ops.add(node.op_type)
        if node.op_type=="Conv" and len(node.input)>=2:
            os_=shapes.get(node.output[0],[]); ws=shapes.get(node.input[1],[])
            if len(os_)==4 and len(ws)==4:
                _,co,ho,wo=os_; _,ci,kh,kw=ws; macs+=co*ci*kh*kw*ho*wo
        elif node.op_type in ("Gemm","MatMul") and len(node.input)>=2:
            sa=shapes.get(node.input[0],[]); sb=shapes.get(node.input[1],[])
            if len(sa)>=2 and len(sb)>=2: macs+=math.prod(sa[:-1])*sa[-1]*sb[-1]
    if BANNED & ops: return None
    return params + nbytes + macs

def score_from_cost(cost):
    if cost is None: return 0.0
    return max(1.0, 25.0 - math.log(max(1, cost))) if cost > 0 else 25.0

# ── Load all models from both datasets ─────────────────────
all_models = {}   # task_name -> {label: (raw, cost)}

for label, root in SOURCE_ZIPS.items():
    if not root.exists():
        print(f"  [!] NOT FOUND: {root}")
        continue
    # Look for zip files or direct onnx files
    zips  = list(root.rglob("*.zip"))
    onnxs = list(root.rglob("task*.onnx"))
    
    loaded = 0
    if zips:
        for zp in zips:
            try:
                with zipfile.ZipFile(zp) as zf:
                    for entry in zf.namelist():
                        name = Path(entry).name
                        if name.startswith("task") and name.endswith(".onnx"):
                            stem = Path(entry).stem
                            raw  = zf.read(entry)
                            cost = compute_cost(raw)
                            if cost is not None:
                                if stem not in all_models:
                                    all_models[stem] = {}
                                all_models[stem][label] = (raw, cost)
                                loaded += 1
            except Exception as e:
                print(f"  [!] zip error {zp}: {e}")
    elif onnxs:
        for p in onnxs:
            stem = p.stem
            raw  = p.read_bytes()
            cost = compute_cost(raw)
            if cost is not None:
                if stem not in all_models:
                    all_models[stem] = {}
                all_models[stem][label] = (raw, cost)
                loaded += 1
    
    print(f"  Loaded [{label}]: {loaded} valid models")

print(f"  Unique tasks across all sources: {len(all_models)}")

# ── Pick best (lowest cost) model per task ─────────────────
best   = {}   # task_name -> raw_bytes
costs  = {}   # task_name -> cost
source = Counter()

for task_name, candidates in all_models.items():
    # Sort by cost ascending → lowest cost = highest score
    ranked = sorted(candidates.items(), key=lambda x: x[1][1])
    winner_label, (winner_raw, winner_cost) = ranked[0]
    best[task_name]  = winner_raw
    costs[task_name] = winner_cost
    source[winner_label] += 1

# ── Add identity for any missing tasks ─────────────────────
from onnx import helper, TensorProto
def make_identity():
    X = helper.make_tensor_value_info("input",  TensorProto.FLOAT, [1,10,30,30])
    Y = helper.make_tensor_value_info("output", TensorProto.FLOAT, [1,10,30,30])
    g = helper.make_graph([helper.make_node("Identity",["input"],["output"])],
                          "id", [X], [Y])
    return helper.make_model(g, opset_imports=[helper.make_opsetid("",17)]).SerializeToString()

missing = []
for i in range(1, NUM_TASKS + 1):
    name = f"task{i:03d}"
    if name not in best:
        best[name]  = make_identity()
        costs[name] = 0   # identity costs 0 but scores 25 if correct, 1 if wrong
        source["identity_fallback"] += 1
        missing.append(name)

if missing:
    print(f"  [!] Missing tasks filled with identity: {missing}")

# ── Pack submission.zip ─────────────────────────────────────
with zipfile.ZipFile(OUT_ZIP, "w", zipfile.ZIP_DEFLATED) as zf:
    for name in sorted(best.keys()):
        num = name.replace("task","")
        zf.writestr(f"task{num}.onnx", best[name])

# ── Score estimate ──────────────────────────────────────────
est_score = sum(
    score_from_cost(costs[n]) for n in best
)

zip_kb = OUT_ZIP.stat().st_size / 1024

print()
print("=" * 55)
print(f"  Tasks in zip       : {len(best)} / {NUM_TASKS}")
print(f"  Estimated score    : {est_score:.2f}")
print(f"  ZIP size           : {zip_kb:.1f} KB  (limit: 1475 KB)")
print()
print("  Source breakdown:")
for lbl, cnt in source.most_common():
    score_contrib = sum(score_from_cost(costs[n]) for n, src in
                        [(n, max(all_models.get(n,{}).items(),
                                 key=lambda x: -x[1][1],
                                 default=(lbl,None))[0])
                         for n in best] if src == lbl)
    print(f"    {lbl:20s}: {cnt:3d} tasks")
print()
print("  Top 10 cheapest tasks (highest score):")
for name in sorted(best, key=lambda n: costs[n])[:10]:
    print(f"    {name}: cost={costs[name]:>10,}  score={score_from_cost(costs[name]):.2f}")
print()
print(f"  ✓ submission.zip saved to {OUT_ZIP}")
print()
print("  Next steps:")
print("  1. Output tab → download submission.zip")
print("  2. Competition → Submit Prediction → File Upload")
print("  3. Upload → Submit")
print("=" * 55)
````
